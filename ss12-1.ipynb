- 以下の方程式を**解析的に**解いてみます。
$$ y= 5x + 3$$
$$ y = 108$$



```python
y=108
x=(y-3)/5
x
```

- **数値シミュレーション**で解いてみます。

```python
x=0
y=108

while(True):
    if(5*x+3>=y):
        print(x)
        print("が答えです。")
        break;
    else:
        print(x)
        print("は答えではありません。")
    x=x+1
    time.sleep(1)
```

- 自然界の方程式はほぼ解くことができません。この場合に有効なのが数値シミュレーションです。
- 円周率πを求める方法もそうです。（モンテカルロ法）
- 実行回数Nを変更してみてください。
- 数式で解けない場合に総当たりで解を探せることが、コンピュータの最も有効な点です。

```python
point = 0
N = 1

for i in range(N):
    x = random.random()
    y = random.random()

    if x*x+y*y < 1.0:
       point += 1
       plt.plot(x, y,"ro")
    else:
       plt.plot(x, y,"bo")
    
# 結果を表示 
plt.show()
pi = 4.0 * point / N
print(pi)
```
