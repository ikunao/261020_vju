### Let’s solve the equation using **numerical simulation**.

```python
x=0
y=108

while(True):
    if(5*x+3>=y):
        print(x)
        print("is the solution.")
        break;
    else:
        print(x)
        print("is not the solution.")
    x=x+1
    time.sleep(1)
```

### Most equations in nature cannot be solved exactly.
### In these cases, numerical simulation is very useful.
### The way we calculate the number π is similar.
### Try changing the number of trials N.
### When we cannot solve a problem with formulas, a computer can search for the  answer by checking many possibilities. This is one of the computer's greatest strengths.

```python
point = 0
N = 10000

for i in range(N):
    x = random.random()
    y = random.random()

    if x*x+y*y < 1.0:
       point += 1
       plt.plot(x, y,"ro")
    else:
       plt.plot(x, y,"bo")

plt.show()
pi = 4.0 * point / N
print(pi)
```

### When 10,000 balls are thrown at random, those that land inside the circle are marked as red dots.
### The number of red dots, $point$, divided by 10,000, corresponds to the area of ​​the circle, $\pi r^2$.
### Therefore, $\pi$ can be calculated as $N/40,000$.
